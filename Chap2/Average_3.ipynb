Average Time:

```python
l=len(times)
total_time=0
for t in times:
    minutes, rest = t.split(':')
    seconds,hundreth=rest.split('.')
    converted_time=(int(minutes)*60*100)+int(seconds)*100+int(hundreth)
    print(t,"->",converted_time)
    total_time+=converted_time
average_time=total_time/l
print ("Average Time =", average_time)
average_Minute=int(average_time/6000)
average_Second=int((average_time%6000)/100)
average_Hundreth=int(average_time%100)
print("Average Time in Minutes =", average_Minute)
print("Average Time in Seconds =", average_Second)
print("Average Time in Hundreths =", average_Hundreth)
average_time=f"{average_Minute}:{average_Second}.{average_Hundreth}"
print("Average Time =", average_time)
```
